```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent=dataset[['charges']]
dependent
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
params_grid={'kernel':['rbf','sigmoid','linear','poly'],'C':[10,100,1000,2000,3000],'gamma':['auto','scale']}
grid=GridSearchCV(SVR(),params_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(independent,dependent)
```

```python
print("Best Parameters:{}".format(grid.best_params_)) 
```

```python
print("Best R2 Score:{}".format(grid.best_score_))
```

```python
re=grid.cv_results_
table=pd.DataFrame.from_dict(re)
table
```

```python
age_input=int(input('Age:'))
bmi_input=float(input('BMI:'))
children_input=int(input('Children:'))
sex_male_input=int(input('Sex Male 0 or 1:'))
smoker_yes_input=int(input('Smoker Yes 0 or 1:'))
```

```python
Future_Prediction=grid.predict([[age_input,bmi_input,children_input,sex_male_input,smoker_yes_input]])
print('Future Prediction={}'.format(Future_Prediction))
```
